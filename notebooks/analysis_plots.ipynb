## Imports

```
# Import python packages
import streamlit as st
import pandas as pd
import altair as alt

# We can also use Snowpark for our analyses!
from snowflake.snowpark.context import get_active_session
session = get_active_session()
```

## Global Analysis

### Number of Trending Videos by Category Over Time

```
df = session.sql("""
    SELECT 
    *
    FROM gold.v_yt_global_trending_videos_count;
""").toPandas()
df.columns = df.columns.str.lower()
df['trending_month'] = pd.to_datetime(df['trending_month'])
chart = alt.Chart(df).mark_line().encode(
    x=alt.X('trending_month:T', title='Month'),
    y=alt.Y('trending_videos_count:Q', title='Count of Trending Videos'),
    color=alt.Color('category_title:N', title='Category')
).properties(
    title='Number of Trending Videos by Category Over Time',
    width=1200
).interactive()

chart
```

### Average Views by Category Over Time

```
df = session.sql("""
    SELECT 
    *
    FROM gold.v_yt_global_avg_views;
""").toPandas()
df.columns = df.columns.str.lower()
df['trending_month'] = pd.to_datetime(df['trending_month'])
chart = alt.Chart(df).mark_line(point=True).encode(
    x=alt.X('trending_month:T', title='Month'),
    y=alt.Y('avg_views:Q', title='Average Views'),
    color=alt.Color('category_title:N', title='Category')
).properties(
    title='Average Views by Category Over Time',
    width=1200
).interactive()
chart
```

## Countries Analysis

### Average Views for Top Categories Over Time by Country

```
df = session.sql("""
    SELECT
    *
    FROM gold.v_yt_country_avg_views;
""").toPandas()
df.columns = df.columns.str.lower()
df['trending_month'] = pd.to_datetime(df['trending_month'])
chart_grid = alt.Chart(df).mark_line(point=True).encode(
    x=alt.X('trending_month:T', title='Month'),
    y=alt.Y('avg_views:Q', title='Average Views'),
    color=alt.Color('category_title:N', title='Category'),
    tooltip=['country', 'trending_month', 'category_title', 'avg_views']
).properties(
    width=350,
    height=250
).facet(
    facet='country:N',
    columns=3
).properties(
    title='Average Views for Top Categories Over Time by Country'
).resolve_scale(
    y='independent'
)

chart_grid
```
